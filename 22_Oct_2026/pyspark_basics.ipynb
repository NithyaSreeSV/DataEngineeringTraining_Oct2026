```python
from pyspark.sql.types import (
    StructType,
    StructField,
    IntegerType,
    StringType,
    DoubleType,
    BooleanType
)
```

```python
data = [
    (101, "Laptop", "Electronics", 65000.0, True),
    (102, "Mouse", "Accessories", 1500.0, True),
    (103, "Monitor", "Electronics", 18000.0, True),
    (104, "Office Chair", "Furniture", 9000.0, False),
    (105, "Keyboard", "Accessories", 3000.0, True)
]
```

```python
schema = StructType([
    StructField("product_id", IntegerType(), False),
    StructField("product_name", StringType(), False),
    StructField("category", StringType(), True),
    StructField("price", DoubleType(), True),
    StructField("available", BooleanType(), True)
])
```

```python
df = spark.createDataFrame(data, schema)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col
```

```python
df2 = df.withColumn(
  "price_with_tax",
  col("price") * 1.18
)

df2.show()
```

```python
df2 = df.withColumn(
  "price",
  col("price") * 1.10
)

df2.show()
```

```python
df2 = df.withColumnRenamed(
"product_name",
"name"
)

df2.show()
```

```python
df2 = df.drop("available")

df2.show()
```

```python
from pyspark.sql.functions import lit

df2 = df.withColumn(
"country",
lit("India")
)

df2.show()
```
