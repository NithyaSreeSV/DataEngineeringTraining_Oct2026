Exercise Set 4 — withColumn()

23. Create: gross_amount using: quantity × unit_price and
Add the gross_amount column.
24. Create: discount_amount using: gross_amount × discount_pct / 100
25. Create: final_amount using: gross_amount - discount_amount
26. Add a constant column: country = India
27. Add: high_value where:
final_amount >= 50000 → Yes
otherwise → No

```python
from pyspark.sql.functions import col, lit, when
```

```python
df = df.withColumn("gross_amount", col("quantity") * col("unit_price"))
```

```python
df = df.withColumn("discount_amount", col("gross_amount") * col("discount_pct") / 100)
```

```python
df = df.withColumn("final_amount", col("gross_amount") - col("discount_amount"))
```

```python
df = df.withColumn("country", lit("India"))
```

```python
df = df.withColumn("high_value", when(col("final_amount") >= 50000, "Yes").otherwise("No"))
```

Exercise Set 5 — Conditional Columns

28. Create a column: order_size using:
final_amount >= 50000 → High
final_amount >= 15000 → Medium
otherwise → Low
29. Create another column: quantity_type where:
quantity >= 4 → Bulk
otherwise → Regular
30. Create a delivery-status category where:
Delivered → Completed
Cancelled/Returned → Failed
Others → In Progress

```python
df = df.withColumn("order_size", when(col("final_amount") >= 50000, "High").when(col("final_amount") >= 15000, "Medium").otherwise("Low"))
```

```python
df = df.withColumn("quantity_type", when(col("quantity") >= 4, "Bulk").otherwise("Regular"))
```

```python
df = df.withColumn("delivery_status", when(col("status") == "Delivered", "Completed")
  .when(col("status").isin("Cancelled", "Returned"), "Failed")
  .otherwise("In Progress"))
```
