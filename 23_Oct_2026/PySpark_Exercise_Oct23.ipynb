```python
df2 = df.withColumn(
  "gross_amount",
  col("quantity") * col("unit_price")
)
df2.select(
  sum("gross_amount").alias("total_gross_sales")
).show()
```

```python
df2 = df.withColumn(
    "discount_amount",
    col("gross_amount") * col("discount_pct") / 100
)
df2.select(
    sum("discount_amount").alias("total_discount_given")
).show()
```

```python
df = df.withColumn(
  "final_amount",
  col("gross_amount") - col("discount_amount")
)
df2.select(
    sum("final_amount").alias("total_final_sales")
).show()
```

```python
df.filter(col("order_status") == "Delivered") \
  .select(sum("final_amount").alias("total_delivered_sales")) \
  .show()
```

```python
df.filter(col("order_status") == "Delivered") \
  .select(avg("final_amount").alias("avg_delivered_value")) \
  .show()
```

```python
df.filter(col("category") == "Electronics") \
  .select(count("*").alias("electronics_count")) \
  .show()
```

```python
df.filter(col("order_status") == "Cancelled") \
  .select(count("*").alias("cancelled_count")) \
  .show()
```

Joins Practice:

```python
customers_data = [
    ("C00001", "Aman Retail", "Hyderabad", "Corporate"),
    ("C00002", "Blue Mart", "Mumbai", "Retail"),
    ("C00003", "Fresh Basket", "Bangalore", "Retail"),
    ("C00004", "Prime Stores", "Delhi", "SMB"),
    ("C00005", "Quick Shop", "Pune", "Corporate")
]

customers_columns = [
    "customer_id",
    "customer_name_master",
    "customer_city",
    "segment"
]

customers_df = spark.createDataFrame(
    customers_data,
    customers_columns
)

customers_df.show()
```

```python
joined_df = df. join(
  customers_df,
  on="customer_id",
  how="inner"
)

joined_df.show()
```

```python
left_join_df = df.join(
  customers_df,
  on="customer_id",
  how="left"
)

left_join_df.show()
```

```python
right_join_df = df.join(
  customers_df,
  on="customer_id",
  how="right"
)

right_join_df.show()
```

```python
full_join_df = df.join(
  customers_df,
  on="customer_id",
  how="outer"
)
full_join_df.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import (
    col,
    row_number,
    rank,
    dense_rank,
    sum,
    avg,
    lag,
    lead
)
```

```python
product_df=df.groupBy("Category","product_name") \
.agg(avg("unit_price").alias("avg_price"))
product_df.show()
```

```python
window_spec = Window.partitionBy("category") \
    .orderBy(col("avg_price").desc())
```

```python
ranked_products = product_df.withColumn(
    "rank",
    row_number().over(window_spec)
)

ranked_products.show()
```

```python
from pyspark. sql. functions import col, to_date

df2 = df.withColumn(
  "order_date",
  to_date(col("order_date"), "yyyy-MM-dd")
)
```

```python
from pyspark.sql.functions import year

df2.withColumn(
  "year",
  year("order_date")
  ).select(
  "order_date",
  "year"
).show()
```

```python
from pyspark.sql. functions import month

df2.withColumn(
  "month",
  month("order_date")
  ).select(
  "order_date",
  "month"
).show()
```

```python
from pyspark.sql. functions import dayofmonth

df2.withColumn(
  "day",
  dayofmonth("order_date")
  ).select(
  "order_date",
  "day"
).show()
```
