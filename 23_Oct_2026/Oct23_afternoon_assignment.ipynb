when() / otherwise():

Create:

1. Efficiency Category

        >= 100% → Above Target
        >= 95% → On Target
        >= 85% → Below Target
        otherwise → Critical

2. Defect Category

        0 → Perfect
        1–10 → Normal
        11–25 → Attention
        above 25 → High Defect

3. Machine Condition

Use temperature and vibration:

        High temperature OR high vibration → Inspect
        otherwise → Normal

Then count records in each resulting category.

```python
df_efficiency_category = df.withColumn("production_efficiency", (col("units_produced") / col("units_planned")) * 100)

df_efficiency_category = df_efficiency_category.withColumn(
    "efficiency_category",
    when(
        col("production_efficiency") >= 100,
        "Above Target"
    ).when(
        col("production_efficiency") >= 95,
        "On Target"
    ).when(
        col("production_efficiency") >= 85,
        "Below Target"
    ).otherwise("Critical")
)

df_efficiency_category.select("production_id", "production_efficiency", "efficiency_category").show()
```

```python
df_defect_category = df.withColumn(
    "defect_category",
    when(
        col("defective_units") == 0,
        "Perfect"
    ).when(
        col("defective_units") <= 10,
        "Normal"
    ).when(
        col("defective_units") <= 25,
        "Attention"
    ).otherwise("High Defect")
)

df_defect_category.select("production_id", "defective_units", "defect_category").show()
```

```python
df_machine_category = df.withColumn(
    "machine_condition",
    when(
        (col("temperature_c") > 80) | (col("vibration_mm_s") > 0.5), "Inspect"
    ).otherwise("Normal")
)

df_machine_category.select("production_id", "temperature_c", "vibration_mm_s", "machine_condition").show()
```

```python
print("Efficiency Categories:")
df_efficiency_category.groupBy("efficiency_category").count().show()

print("Defect Categories:")
df_defect_category.groupBy("defect_category").count().show()

print("Machine Condition Categories:")
df_machine_category.groupBy("machine_condition").count().show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import (row_number,rank,dense_rank,sum,avg,lag,lead,col)
```

```python
window_plant = Window.partitionBy("plant_id").orderBy(col("units_produced").desc())

df_rank = df.withColumn("production_rank",rank().over(window_plant))

df_rank.select("plant_id","production_id","units_produced","production_rank").show()
```

```python
df_row_number = df.withColumn("row_number", row_number().over(window_plant))

df_row_number.select("plant_id","production_id","units_produced","row_number").show()
```

```python
df_rank = df.withColumn("rank",rank().over(window_plant))

df_rank.select("plant_id","production_id","units_produced","rank").show()
```

```python
df_dense_rank = df.withColumn("dense_rank",dense_rank().over(window_plant))

df_dense_rank.select("plant_id","production_id","units_produced","dense_rank").show()
```

```python
df_top5 = df.withColumn("row_number", row_number().over(window_plant)).filter(col("row_number") <= 5)

df_top5.show()
```

```python
df_machine_output = df.groupBy("plant_id","machine_id").agg(sum("units_produced").alias("total_output"))

window_machine = Window.partitionBy("plant_id").orderBy(col("total_output").desc())

df_top_machines = df_machine_output.withColumn("machine_rank",dense_rank().over(window_machine)).filter(col("machine_rank") <= 3)

df_top_machines.show()
```

```python
window_plant_all = Window.partitionBy("plant_id")

df_plant_total = df.withColumn("total_plant_production",sum("units_produced").over(window_plant_all))

df_plant_total.select("plant_id","production_id","units_produced","total_plant_production").show()
```

```python
window_machine_all = Window.partitionBy("machine_id")

df_machine_avg = df.withColumn("average_machine_output",avg("units_produced").over(window_machine_all))

df_machine_avg.select("machine_id","production_id","units_produced","average_machine_output").show()
```

```python
window_cumulative = Window.partitionBy("machine_id").orderBy("production_date").rowsBetween(Window.unboundedPreceding,Window.currentRow)

df_cumulative = df.withColumn("cumulative_production",sum("units_produced").over(window_cumulative))

df_cumulative.select("machine_id","production_date","units_produced","cumulative_production").show()
```

```python
window_date = Window.partitionBy("machine_id").orderBy("production_date")

df_lag = df.withColumn("previous_output",lag("units_produced", 1).over(window_date))

df_lag.select("machine_id","production_date","units_produced","previous_output").show()
```

```python
df_lead = df.withColumn("next_output",lead("units_produced", 1).over(window_date))

df_lead.select("machine_id","production_date","units_produced","next_output").show()
```

```python
df_change = df.withColumn("previous_output",lag("units_produced", 1).over(window_date))

df_change = df_change.withColumn("output_change",col("units_produced") - col("previous_output"))

df_change.select("machine_id","production_date","units_produced","previous_output","output_change").show()
```

```python
df_family_output = df.groupBy("product_family").agg(sum("units_produced").alias("total_output"))

window_family = Window.orderBy(col("total_output").desc())

df_family_rank = df_family_output.withColumn("family_rank",dense_rank().over(window_family))

df_family_rank.show()
```

```python
from pyspark.sql.functions import (
    to_date, year, month, dayofmonth, date_format,
    sum, col, desc, date_add, date_sub, datediff, current_date
)
base_df = spark.read.csv("manufacturing_production_120k.csv", header=True, inferSchema=True)
```

```python
df = base_df.withColumn("production_date", to_date(col("production_date")))
df.select("production_date").printSchema()
```

```python
df.select("production_date", year("production_date").alias("year")).show(5)
```

```python
df.select("production_date", month("production_date").alias("month")).show(5)
```

```python
df.select("production_date", dayofmonth("production_date").alias("day")).show(5)
```

```python
df.select("production_date", date_format("production_date", "MMMM").alias("month_name")).show(5)
```

```python
df.select("production_date", date_format("production_date", "EEEE").alias("day_name")).show(5)
```

```python
df = df.withColumn("year_month", date_format("production_date", "yyyy-MM"))
df.select("production_date", "year_month").show(5)
```

```python
df.groupBy("year_month").agg(sum("units_produced").alias("monthly_production")).orderBy("year_month").show()
```

```python
df.groupBy("year_month").agg(sum("defective_units").alias("monthly_defects")).orderBy("year_month").show()
```

```python
df.groupBy("year_month", "plant_id").agg(sum("units_produced").alias("total_units")).orderBy("year_month", "plant_id").show()
```

```python
df.filter(col("production_date").between("2023-01-01", "2023-03-31")).show(5)
```

```python
df.select("production_date", date_add("production_date", 7).alias("plus_7_days")).show(5)
```

```python
df.select("production_date", date_sub("production_date", 30).alias("minus_30_days")).show(5)
```

```python
df.select("production_date", datediff(current_date(), col("production_date")).alias("days_ago")).show(5)
```

```python
df.groupBy("year_month").agg(sum("units_produced").alias("total_output")).orderBy(desc("total_output")).limit(1).show()
```
